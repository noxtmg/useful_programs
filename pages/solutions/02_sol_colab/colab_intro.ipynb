## Różniczkowanie

### 1. Różniczka funkcji jednej zmiennej $y = f(x)$

Wzór ogólny na różniczkę funkcji jednej zmiennej:

$$
dy = f'(x) \cdot dx
$$

---

### Przykład: Wyznaczenie różniczki funkcji $y = x^3 + 2x$

1. **Definicja funkcji:**

   $$
   f(x) = x^3 + 2x
   $$

2. **Obliczenie pochodnej $f'(x)$:**

   $$
   f'(x) = \frac{d}{dx}(x^3 + 2x) = 3x^2 + 2
   $$

3. **Złożenie wzoru na różniczkę $dy$:**

   $$
   dy = (3x^2 + 2) \, dx
   $$

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML, display

# 1. Konfiguracja fizyki wahadła i parametrów czasowych
L = 1.0         # długość wahadła [m]
g = 9.81        # przyspieszenie ziemskie [m/s^2]
theta0 = np.radians(45)  # kąt początkowy (45 stopni)
omega = np.sqrt(g / L)   # częstość kołowa drgań

fps = 30
duration = 4    # czas trwania w sekundach
frames = fps * duration
t = np.linspace(0, duration, frames)

theta = theta0 * np.cos(omega * t)

x = L * np.sin(theta)
y = -L * np.cos(theta)

fig, ax = plt.subplots(figsize=(6, 6))
ax.set_xlim(-1.2 * L, 1.2 * L)
ax.set_ylim(-1.2 * L, 0.2 * L)
ax.set_aspect('equal')
ax.grid(True, linestyle='--', alpha=0.6)
ax.set_title('Animacja Wahadła Prostego', fontsize=14)
ax.set_xlabel('x [m]')
ax.set_ylabel('y [m]')

ax.plot(0, 0, 'go', markersize=8, label='Punkt zawieszenia')
line, = ax.plot([], [], 'o-', lw=2, color='navy', markersize=12)

def init():
    line.set_data([], [])
    return line,

def animate(i):
    line.set_data([0, x[i]], [0, y[i]])
    return line,

ani = animation.FuncAnimation(
    fig, animate, init_func=init,
    frames=frames, interval=1000/fps, blit=True
)

# Zapis animacji do pliku video.mp4 na dysku
ani.save('video.mp4', writer='ffmpeg', fps=fps)

plt.close(fig)

# Wyświetlenie animacji jako wideo HTML5 bezpośrednio w Colabie
display(HTML(ani.to_html5_video()))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 1. Nowy parametr t i zmienione równania parametryczne (Krzywa Vivianiego)
t = np.linspace(0, 4 * np.pi, 1000)
a = 0.75

x = a * (1 + np.cos(t))
y = a * np.sin(t)
z = 2 * a * np.sin(t / 2)

# 2. Tworzenie wykresu 3D
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')

# Rysowanie nowej krzywej parametrycznej
ax.plot(x, y, z, color='crimson', linewidth=2, label="Krzywa Vivianiego")

# Stylizacja i opis osi
ax.set_title("3D Parametric Curve: Krzywa Vivianiego", fontsize=14)
ax.set_xlabel("x(t)")
ax.set_ylabel("y(t)")
ax.set_zlabel("z(t)")
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(loc="upper right")

# 3. Zapis wykresu do pliku parametric_curve.png (PRZED plt.show())
plt.savefig("parametric_curve.png", dpi=150, bbox_inches="tight")

# 4. Wyświetlenie wykresu w notebooku
plt.show()
```
